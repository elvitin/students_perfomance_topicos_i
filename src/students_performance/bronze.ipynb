# Camada Bronze — Students Performance

**Arquitetura Medalhão — Camada Bronze**

Responsabilidade: ingestão bruta dos dados originais sem nenhum tratamento.  
Os dados são armazenados exatamente como foram recebidos, preservando fidelidade total à fonte.

**Fonte:** `data/student_performance_dataset.csv`  
**Origem:** https://www.kaggle.com/datasets/rabieelkharoua/students-performance-dataset  
**Destino:** tabela `bronze_students` no SQLite (`data_lakehouse.db`)

```python
import sqlite3
import pandas as pd
```

```python
# Conexão com o banco de dados SQLite (lakehouse local)
conn = sqlite3.connect("../../data/data_lakehouse.db")
```

## 1. Leitura do CSV bruto

```python
# Leitura do dataset original — sem qualquer transformação
students_df = pd.read_csv("../../data/student_performance_dataset.csv")

print(f"Shape: {students_df.shape}")
print(f"Colunas: {list(students_df.columns)}")
```

## 2. Inspeção inicial

```python
# Primeiras linhas do dataset
students_df.head(10)
```

```python
# Tipos de dados e valores nulos
students_df.info()
```

```python
# Estatísticas descritivas básicas
students_df.describe()
```

```python
# Verificação de valores nulos por coluna
nulos = students_df.isnull().sum()
print("Valores nulos por coluna:")
print(nulos[nulos >= 0].to_string())
```

```python
# Verificação de duplicatas
n_duplicatas = students_df.duplicated().sum()
print(f"Total de linhas duplicadas: {n_duplicatas}")

n_dup_id = students_df.duplicated(subset=["StudentID"]).sum()
print(f"StudentIDs duplicados: {n_dup_id}")
```

## 3. Persistência na camada Bronze

```python
# Gravação na tabela bronze_students — dados brutos, sem modificações
rows = students_df.to_sql("bronze_students", conn, if_exists="replace", index=False)
print(f"Registros gravados em bronze_students: {rows}")
```

```python
# Validação: confirmar que a tabela foi criada corretamente
count = pd.read_sql("SELECT COUNT(*) as total FROM bronze_students", conn)
print(f"Total de registros em bronze_students: {count['total'][0]}")
```

```python
conn.close()
print("Camada Bronze concluída com sucesso.")
```
