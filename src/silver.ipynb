```python
import sqlite3
import pandas as pd
```

```python
# Conexão com o banco de dados SQLite
conn = sqlite3.connect("../data/data_lakehouse.db")
```

```python
clientes_df = pd.read_sql("SELECT * FROM bronze_clientes", conn)
produtos_df = pd.read_sql("SELECT * FROM bronze_produtos", conn)
vendas_df = pd.read_sql("SELECT * FROM bronze_vendas", conn)
```

```python
# Limpeza
vendas_df = vendas_df[vendas_df["valor"] > 0]
vendas_df["data"] = pd.to_datetime(vendas_df["data"])

clientes_df = clientes_df.drop_duplicates()
produtos_df = produtos_df.drop_duplicates()
```

```python
clientes_df.to_sql("silver_clientes", conn, if_exists="replace", index=False)
produtos_df.to_sql("silver_produtos", conn, if_exists="replace", index=False)
vendas_df.to_sql("silver_vendas", conn, if_exists="replace", index=False)
```
