```python
import sqlite3
from pathlib import Path
import pandas as pd
```

```python
DB_PATH = Path("../../data/data_lakehouse.db").resolve()
conn = sqlite3.connect(DB_PATH)

dim_aluno = pd.read_sql("SELECT * FROM dim_aluno", conn)
fato_desempenho = pd.read_sql("SELECT * FROM fato_desempenho", conn)

df = dim_aluno.merge(fato_desempenho, on="StudentID", how="inner")

df.head()
```

```python
dim_aluno.head()
```

```python
fato_desempenho.head()
```

```python
df
```

```python

df=pd.get_dummies(df,columns=['ParentalEducation_label'],drop_first=True)
df.head()
```

```python
df=pd.get_dummies(df,columns=['ParentalSupport_label'], drop_first=True)
df.head()
```

```python
df=pd.get_dummies(df, columns=['Age'], drop_first=True)
df
```

```python
df['AbscencesGroup'] = pd.cut(df['Absences'], bins=[0,10,20,30], labels=['Baixa','Media','Alta'], include_lowest=True)
df
```

```python
df = pd.get_dummies(df,columns=['AbscencesGroup'], drop_first=True)
df
```

```python
df['StudyGroup'] = pd.cut(df['StudyTimeWeekly'], bins=[0,5,10,15,20], labels=['Muito baixo', 'Baixa', 'Média', 'Alta'], include_lowest=True)
df
```

```python
df = pd.get_dummies(df, columns=['StudyGroup'], drop_first=True)
df
```

```python
print(df.columns)
```

```python
df = df.drop(columns={
    'StudentID',
    'Gender_label',
    'ParentalEducation',
    'Absences',
    'ParentalSupport',
    'Tutoring_label',
    'Extracurricular_label',
    'Sports_label',
    'Music_label',
    'Volunteering_label',
    'GradeClass_label',
    'StudyTimeWeekly',
    'GPA'
})

df.head()
```

```python
print(df.columns)
```

```python
bool_cols = df.select_dtypes(include='bool').columns

df[bool_cols].head()
```

```python
df[bool_cols] = df[bool_cols].astype(int)
df
```
