```python
import sqlite3
import pandas as pd
```

```python
# Conexão com o banco de dados SQLite
conn = sqlite3.connect("../../data/clients_products_sells_dataset/data_lakehouse.db")
```

```python
clientes_df = pd.read_csv("../../data/clients_products_sells_dataset/clientes.csv")
produtos_df = pd.read_csv("../../data/clients_products_sells_dataset/produtos.csv")
vendas_df = pd.read_csv("../../data/clients_products_sells_dataset/vendas.csv")
```

```python
clientes_df.to_sql("bronze_clientes", conn, if_exists="replace", index=False)
produtos_df.to_sql("bronze_produtos", conn, if_exists="replace", index=False)
vendas_df.to_sql("bronze_vendas", conn, if_exists="replace", index=False)
```

```python
conn.close()
```
