```python
import sqlite3
import pandas as pd
```

```python
# Conexão com o banco de dados SQLite
conn = sqlite3.connect("../../data/clients_products_sells_dataset/data_lakehouse.db")
```

```python
clientes_df = pd.read_sql("SELECT * FROM silver_clientes", conn)
produtos_df = pd.read_sql("SELECT * FROM silver_produtos", conn)
vendas_df = pd.read_sql("SELECT * FROM silver_vendas", conn)
```

```python
# Dimensões
clientes_df.to_sql("dim_cliente", conn, if_exists="replace", index=False)
produtos_df.to_sql("dim_produto", conn, if_exists="replace", index=False)

# Fato
vendas_df.to_sql("fato_vendas", conn, if_exists="replace", index=False)
```

```python
# Tabela agregada
query = """
SELECT 
    v.data,
    p.categoria,
    SUM(v.valor) as receita
FROM silver_vendas v
JOIN silver_produtos p
    ON v.produto_id = p.produto_id
GROUP BY v.data, p.categoria
"""

receita_categoria = pd.read_sql(query, conn)

receita_categoria.to_sql("gold_receita_categoria", conn, if_exists="replace", index=False)
```

```python
conn.close()
```
